## Data transformations

Let's modify data from existing columns to create the new ones.

We can calculate the level of inequality of happiness by substracting the difference between happiness score and its mean and dividing on standard deviation. It shows the level of inequality all over the world as how much happiness in every country deviates from the average happiness score in all countries. 
Besides, let's calculate the trust level to government by its citizens by dividing on 2 the sum of social support score and perceptions of corruption. As a result, this column presents how much government is trusted in each country by considering the most significant factors that people expect from the authorities.

```python
world_happiness_2015['happiness_inequality'] = (world_happiness_2015['happiness_score'] - world_happiness_2015['happiness_score'].mean()) / world_happiness_2015['happiness_score'].std()
world_happiness_2015['trust_level'] = (world_happiness_2015['social_support'] + world_happiness_2015['perceptions_of_corruption']) / 2
world_happiness_2015.head()
```

# 2.1. Overview (with simple plots)

Our main work with dataset is finished. So now I will provide several hypotheses based on my data and build graphs for it to deeply analyze the data and check the validity. 

##### *#1 hypothesis: the higher level of GDP per capita provides the higher happiness score.* 

I will analyze it in the context of Western Europe as it has the most developed economy and these countries are placed on top of the table. 

There I decided to build a scatter plot using Plotly to make it more interactive. Size of circles depends on the level of GDP per capita and color depends on level of happiness, also by animations in plotly it's possible to move to different years. 

```python
world_happiness_2015['year'] = 2015
world_happiness_2016['year'] = 2016
world_happiness_2017['year'] = 2017
world_happiness_2018['year'] = 2018
world_happiness_2019['year'] = 2019
world_happiness_2020['year'] = 2020
world_happiness_2021['year'] = 2021
world_happiness_2022['year'] = 2022
world_happiness_2023['year'] = 2023

concatted = pd.concat(
    [world_happiness_2015, world_happiness_2016, world_happiness_2017, world_happiness_2018, world_happiness_2019,
     world_happiness_2020, world_happiness_2021, world_happiness_2022, world_happiness_2023], ignore_index=True)

fig = px.scatter(concatted[concatted['region'] == 'Western Europe'],
                 x='gdp_per_capita', y='happiness_score', text='country', size='gdp_per_capita',
                 color='happiness_score', animation_frame='year',
                 labels={'gdp_per_capita': 'gdp_per_capita', 'happiness_score': 'happiness_score', 'year': 'year'},
                 title='Correlation between GDP per Capita and Happiness Score in Western Europe (2015–2023)',
                 template='plotly_white'
                 )
fig.update_layout(width=1000, height=800)
fig.show()
```

```python
correlation = concatted[['gdp_per_capita', 'happiness_score']].corr()
print(correlation)
```

Here is the scatter plot showing connection between GDP per capita and Happiness level in Western Europe by the size of circles and their shade. To continue, from the graphs it's obvious in each year that circles that smaller (low GDP per capita level) also have a darker shade, so those countries where GDP is low also have less happiness score. Computed correlation also showed the strong dependence between factors. So, hypothesis is proved.

# 2.2. Overview (with more complex plots)

#### *#1 hypothesis: happiness level higher in those countrues where GDP per capita, social support and freedom to make life choices higher respectively.*

For this I decided to construct graph in the form of the world map using Plotly graph objects. It helps to vizualise data more evident as I wanted to provide ability to see all the years via one graph. It's possible here to choose a year of data that you want to see as I concatted data from various tables and created a column for year to do it. 

```python
import plotly.graph_objects as go
import ipywidgets as widgets

world_happiness_2015['year'] = 2015
world_happiness_2016['year'] = 2016
world_happiness_2017['year'] = 2017
world_happiness_2018['year'] = 2018
world_happiness_2019['year'] = 2019
world_happiness_2020['year'] = 2020
world_happiness_2021['year'] = 2021
world_happiness_2022['year'] = 2022
world_happiness_2023['year'] = 2023

concatted = pd.concat([world_happiness_2015, world_happiness_2016, world_happiness_2017, world_happiness_2018, world_happiness_2019, world_happiness_2020, world_happiness_2021, world_happiness_2022, world_happiness_2023], ignore_index=True)

years = [2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023]

def change_year(current_year):
    wh_data = concatted[concatted['year'] == current_year]

    fig = go.Figure(go.Choropleth(
        locations=wh_data['country'],  
        locationmode='country names',  
        z=wh_data['happiness_score'],  
        hoverinfo='location+z+text',  
        text=wh_data['country'],  
        hovertemplate=(
            '<b>%{text}</b><br>'  
            'happiness_score: %{z}<br>'  
            'gdp_per_capita: %{customdata[0]}<br>'  
            'social_support: %{customdata[1]}<br>'  
            'freedom_to_make_life_choices: %{customdata[2]}<br>' 
        ),
        customdata=wh_data[['gdp_per_capita', 'social_support', 'freedom_to_make_life_choices']].values,
        colorscale='pinkyl',
    ))

    fig.update_layout(width=1000, height=600)


    fig.show()

chosen_year = widgets.Dropdown(options=years)

widgets.interactive(change_year, current_year=chosen_year)
```

```python
correlation = concatted[['gdp_per_capita', 'happiness_score', 'social_support', 'freedom_to_make_life_choices']].corr()
print(correlation)
```

Here is the Choropleth graph. From analyzing all the years we can see that pinkier countries (happier ones) also show high indecators of GDP per capita, social support and freedom to make choices. It's logical as those factors are one of the most significant as people are happier when they live in economic developed countries which support them in social ways providing ability to choose what you want. Moreover, computed correlation showed the strong correlation between factors. Hypothesis is proved.

#### *#2 hypothesis: The economic well-being and social support have a positive impact on happiness, and this impact varies depending on the region. Additionally, the strength of this influence increases in countries with higher levels of freedom of choice.*

For this I decided to use Plotly to construct 3D-scatter plot. It will be possible to choose exact year, size of circles depends on happiness level and data presented by regions.

```python
data = concatted.groupby(['year', 'region', 'gdp_per_capita', 'social_support', 'freedom_to_make_life_choices'])[
    'happiness_score'].mean().reset_index()

fig = px.scatter_3d(data, x='gdp_per_capita', y='social_support', z='happiness_score',
                    color='region',
                    size='happiness_score',
                    animation_frame='year',
                    title='Influence of economic situation and social support on happiness level considering citizens freedom',
                    labels={'gdp_per_capita': 'gdp_per_capita',
                            'social_support': 'social_support',
                            'happiness_score': 'happiness_score',
                            'region': 'region'},
                    template='plotly_white', )

fig.update_layout(width=1200, height=700)

fig.show()
```

```python
correlation = concatted[['gdp_per_capita', 'happiness_score', 'social_support']].corr()
print(correlation)
```

Here is the 3D-scatter providing data really visual. So, it's obvious from the graph that regions placed differently and the highest happiness scores also have high levels of social support which provided also high economic well-being. Also computed correlation showed strong connection between factors. Hypothesis is proved.
